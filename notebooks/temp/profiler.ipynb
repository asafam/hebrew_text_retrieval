```python
import json
import os

os.chdir("/Users/asaf/Downloads")

INPUT_PATH = "merged_trace_fixed.json"
OUTPUT_PATH = "merged_trace_fixed2.json"

# Load raw trace
with open(INPUT_PATH) as f:
    events = json.load(f)

# Find the true beginning of trace
ts_values = [e["ts"] for e in events if "ts" in e]
min_ts = min(ts_values)

# Define max allowed ts: e.g. 10 seconds after trace start
MAX_TS = min_ts + 10_000_000_000  # 10M µs = 10,000 sec

# Filter out future events
filtered_events = []
for e in events:
    if "ts" not in e or e["ts"] <= MAX_TS:
        filtered_events.append(e)

# Normalize timestamps
for e in filtered_events:
    if "ts" in e:
        e["ts"] -= min_ts

# Save cleaned trace
with open(OUTPUT_PATH, "w") as f:
    json.dump(filtered_events, f)

print(f"✅ Cleaned trace written to: {OUTPUT_PATH}")
print(f"Kept {len(filtered_events)} of {len(events)} events")
```
